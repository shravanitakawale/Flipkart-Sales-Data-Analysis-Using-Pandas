**Concatenation- its used to merge two different dataframe**

````python
df1=pd.DataFrame({'name':['A','B','C','D','np.Nan'],
                  'age' :[10,11,12,13,14]})
df2=pd.DataFrame({'name':['A','B','C','E','np.NaN'],
                  'marks' :[87,97,25,66,12]})
````

````python
df2
````

**Vertical Concatenation- Its used to concatenate the dataframe row by row**

````python
pd.concat([df1,df2])  #so many Null values
````

**Horizontal Concatenation- its used to concatenate the dataframe column by column**

````python
pd.concat([df1,df2],axis=1)  #duplicate column has been created
````

**Join- Its used to merge the two different dataframe based on the common columns**

**Inner Join**

````python
df1.merge(df2, how='inner',on='name')
````

````python
df1.merge(df2, how='left',on='name')  #common data and left data
````

````python
df1.merge(df2, how='right',on='name')
````

````python
df1.merge(df2, how='outer',on='name')
````
